Implementation of chain matrix multiplication using dynamic
programming.

```python
def matrix_chain_multiplication(dimensions):
    n = len(dimensions) - 1

    # dp[i][j] stores minimum number of multiplications
    # needed to multiply matrices from i to j
    dp = [[0 for _ in range(n)] for _ in range(n)]

    # chain_length is the length of the matrix chain
    for chain_length in range(2, n + 1):

        for i in range(n - chain_length + 1):
            j = i + chain_length - 1

            # Initialize with infinity
            dp[i][j] = float('inf')

            # Try all possible positions to split the chain
            for k in range(i, j):

                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + dimensions[i] * dimensions[k + 1] * dimensions[j + 1]
                )

                # Store minimum cost
                dp[i][j] = min(dp[i][j], cost)

    return dp[0][n - 1]


# Example
dimensions = [10, 20, 30, 40,]

minimum_operations = matrix_chain_multiplication(dimensions)

print("Minimum number of multiplications:", minimum_operations)
```
