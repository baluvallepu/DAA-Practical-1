```python

# Practical 12: Travelling Salesman Problem

from itertools import permutations

def travelling_salesman(graph):
    n = len(graph)
    cities = range(1, n)

    min_cost = float('inf')
    best_path = []

    # Generate all possible orders of visiting cities
    for path in permutations(cities):
        current_path = (0,) + path + (0,)
        cost = 0

        # Calculate total cost of the tour
        for i in range(len(current_path) - 1):
            cost += graph[current_path[i]][current_path[i + 1]]

        # Update minimum cost and best path
        if cost < min_cost:
            min_cost = cost
            best_path = current_path

    print("Optimal Tour:")
    print(" -> ".join(str(city) for city in best_path))
    print("Minimum Travel Cost:", min_cost)


# Cost matrix representing distances between cities
graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

travelling_salesman(graph)
```
