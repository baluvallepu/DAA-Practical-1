```python

# Practical 11: Floyd-Warshall Algorithm

def floyd_warshall(graph):
    n = len(graph)

    # Create a copy of the graph
    dist = [row[:] for row in graph]

    # Find shortest paths between all pairs of vertices
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] != float('inf') and dist[k][j] != float('inf'):
                    dist[i][j] = min(
                        dist[i][j],
                        dist[i][k] + dist[k][j]
                    )

    # Display the shortest distance matrix
    print("Shortest Distance Matrix:")
    for row in dist:
        for value in row:
            if value == float('inf'):
                print("INF", end="  ")
            else:
                print(value, end="    ")
        print()


# Graph representation using an adjacency matrix
# INF represents no direct connection
INF = float('inf')

graph = [
    [0,   5,   INF, 10],
    [INF, 0,   3,   INF],
    [INF, INF, 0,   1],
    [INF, INF, INF, 0]
]

floyd_warshall(graph)
```
