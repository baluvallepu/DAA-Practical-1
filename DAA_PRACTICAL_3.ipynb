## **MAX HEAP SORT**

```python
import time


def max_heap(arr, size, root):
    big = root
    l = (2 * root) + 1
    r = (2 * root) + 2

    if l < size and arr[l] > arr[big]:
        big = l

    if r < size and arr[r] > arr[big]:
        big = r

    if big != root:
        temp = arr[root]
        arr[root] = arr[big]
        arr[big] = temp
        max_heap(arr, size, big)


def max_heap_sort(arr):
    length = len(arr)

    for index in range((length // 2) - 1, -1, -1):
        max_heap(arr, length, index)

    end = length - 1

    while end > 0:
        arr[0], arr[end] = arr[end], arr[0]
        max_heap(arr, end, 0)
        end -= 1


count = int(input("Enter number of elements: "))

numbers = []

print("Enter the elements (space-separated):")
# Read all numbers as a single line, split them, and convert to integers
input_elements_str = input().split()

# Convert each string element to an integer and append to the list
# It's good practice to ensure the number of inputs matches 'count'
# However, to maintain the spirit of the original loop, we'll append all provided.
for element in input_elements_str:
    numbers.append(int(element))

print("\nOriginal Array:", numbers)

start = time.perf_counter()

max_heap_sort(numbers)

finish = time.perf_counter()

print("\nSorted Array:", numbers)

print("\nExecution Time: {:.10f} seconds".format(finish - start))

print("\nTime Complexities:")
print("Best Case   : O(n log n)")
print("Average Case: O(n log n)")
print("Worst Case  : O(n log n)")
```
